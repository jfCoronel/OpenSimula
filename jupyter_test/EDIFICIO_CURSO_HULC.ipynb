## EDIFICIO CURSO ATECYR HULC

Edificio importado desde HULC, es el edificio que usamos como pequeño terciario en los CURSOS de ATECYR.

```python
import opensimula as osm

sim = osm.Simulation()
pro = sim.new_project("pro")
pro.read_json("edificio_curso_hulc.json")
```

```python
import datetime as dt

date = dt.datetime(2001,3,21,13,0,0)
#pro.show_3D(jupyter=True)
pro.show_3D_shadows(date,jupyter=True)
#pro.show_3D_shadows_animation(date,jupyter=True)
```

```python
pro.simulate()
# 30s calculando las sombras de directa
# 4m:55s Simulando con shadow_calculation = INSTANT, n_iter = 4
# 1m:21s Simulando con shadow_calculation = INTERPOLATION, n_iter = 4
```
