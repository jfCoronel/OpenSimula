```python
import opensimula as osm

sim = osm.Simulation()
pro = sim.new_project("pro")
pro.read_json("cases/case_900FF.json")
```

```python
pro.simulate()
```

```python
import numpy as np
import pandas as pd

# zone temperature
temp_med = pro.component("spaces_1").variable("temperature").values.mean()
temp_max = pro.component("spaces_1").variable("temperature").values.max()
temp_min = pro.component("spaces_1").variable("temperature").values.min()
i_temp_max = np.argmax(pro.component("spaces_1").variable("temperature").values)
i_temp_min = np.argmin(pro.component("spaces_1").variable("temperature").values)

print("Average Temp. (ºC): ", temp_med)
print("Min. Temp. (ºC): ", temp_min)
print("Max. Temp. (ºC): ", temp_max)
print("Min. Temp. Date: ",pro.dates()[i_temp_min])
print("Max. Temp. Date: ",pro.dates()[i_temp_max])
```

```python
feb_1_df = pd.DataFrame()
intervalo = ["2001-02-01","2001-02-02"]
aux =  pro.component("spaces_1").variable_dataframe(interval=intervalo)
feb_1_df["T (ºC)"] = aux["temperature"]
feb_1_df.to_clipboard()
feb_1_df
```
