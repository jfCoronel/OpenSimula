```python
project_dict = {
    "name": "First example project",
    "time_step": 3600,
    "n_time_steps": 8760,
    "initial_time": "01/01/2001 00:00:00",
    "components": [
         {
            "type": "File_met",
            "name": "met",
            "file_type": "MET",
            "file_name": "../../mets/sevilla.met"
        },
        {
            "type":"Calculator",
            "name":"conversor",
            "input_variables": ["T = met.temperature", "w = met.abs_humidity"],
            "output_variables": ["T_F","W_kg"],
            "output_units": ["ºF","kg/kg a.s."],
            "output_expressions": ["T * 9/5 + 32", "w / 1000"]
        }
    ]
}
```

```python
import opensimula.Simulation as Simulation

sim = Simulation()
pro = sim.new_project("First example project")
pro.read_dict(project_dict)
pro.simulate()
```

```python
variables = [pro.component("office").variable("f"), pro.component("office").variable("people_density")]
sim.plot(pro.dates(), variables)
```
