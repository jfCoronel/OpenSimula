```python
import pandas as pd
import numpy as np

# Heating and Cooling Loads
load = pro_cp.component("spaces_1").variable("system_sensible_heat").values
annual_heating = np.where(load>0,load,0).sum()/1e6
annual_cooling = np.where(load<0,-load,0).sum()/1e6
peak_heating = load.max()/1000
i_peak_heating = np.argmax(load)
peak_cooling = -load.min()/1000
i_peak_cooling = np.argmin(load)

resul = pd.DataFrame()
resul["Annual_heating"] = [annual_heating]
resul["Annual_cooling"] = [annual_cooling]
resul["Peak_heating"] = [peak_heating]
resul["Peak_cooling"] = [peak_cooling]
resul
```
